```python
import pandas as pd
import glob
```

```python
files=glob.glob('../data/*.xlsx')
files
```

```python
data=pd.read_excel(files[0],skiprows=2,dtype={"Postleitzahlgebiet":str})
```

```python
data.columns
```

```python
data.sort_values(by='Transformationsindex Infrastrukturrelevanz', inplace=True, ascending=False)
data
```

```python
data_top_100=data.head(1000)
data_top_100["Postleitzahlgebiet"]=data_top_100["Postleitzahlgebiet"].astype(str)
data_top_100["join_key"] = data_top_100["Postleitzahlgebiet"].str.zfill(5)
```

```python
import geopandas as gpd
```

```python
plz_shapes=gpd.read_file("/Users/halukamaier-borst/Documents/CodeRelated/HBDataviz_HalukaOnly/2026/01_IWKarte/Wrangle/data/PLZDaten/02_Shape/PLZ_5_Flaeche_region.shp")
plz_shapes
```

```python
plz_shapes = plz_shapes.set_crs(epsg=3044,allow_override=True)

plz_shapes = plz_shapes.to_crs(epsg=4326)
plz_shapes
```

```python
plz_shapes["join_key"] = plz_shapes["Plz_5"].astype(str)
data_top_100["join_key"]  = data_top_100["Postleitzahlgebiet"].astype(str)
data_top_100
```

```python


merged = plz_shapes.merge(data_top_100, on="join_key", how="inner")   # or how="inner" if you only want matches
merged["geometry"] = merged.geometry.simplify(0.001, preserve_topology=True)
merged
```

```python

# (Optional) Reproject before export (GeoJSON often expected in WGS84)

# 4) Write to GeoJSON
merged.to_file("output2.geojson", driver="GeoJSON")
```

```python
dissolved_gdf = merged.dissolve()
dissolved_gdf.to_file("dissolved.geojson", driver="GeoJSON")
```
