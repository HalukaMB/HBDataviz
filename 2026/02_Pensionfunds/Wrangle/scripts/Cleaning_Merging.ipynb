```python
import pandas as pd
import glob
import numpy as np
```

```python
files=glob.glob('../data/*.xlsx')
files
```

```python
excel_file = pd.ExcelFile("../data/Pensionsfonds_2.xlsx")
sheet_names = excel_file.sheet_names
print(sheet_names)
```

```python
df_norway=pd.read_excel(excel_file, sheet_name=sheet_names[1])
df_norway
```

```python
df_columns_with_char = df_norway.columns[df_norway.columns.str.contains(")", na=False, regex=False)]
index_numbers = [df_norway.columns.get_loc(col) for col in df_columns_with_char]
print(index_numbers)
```

```python
df_norway.columns[index_numbers[0]:(index_numbers[0]+3)]
```

```python
norway_red_df=df_norway[df_norway.columns[index_numbers[0]:(index_numbers[0]+3)]]
norway_red_df
```

```python
# Find rows where any cell contains the character
search_string = ")"
mask = norway_red_df.apply(lambda col: col.astype(str).str.contains(search_string, na=False, regex=False), axis=1).any(axis=1)
norway_red_pos = norway_red_df[mask]
norway_red_pos.columns=["Rank","Name","Value"]
norway_red_pos
```

```python
norway_red_pos["Suffix"]=0
norway_red_pos["Suffix"] = np.where(norway_red_pos["Value"].str.contains("BLN"), 10**9,norway_red_pos["Suffix"] )
norway_red_pos["Suffix"] = np.where(norway_red_pos["Value"].str.contains("MLN"), 10**6,norway_red_pos["Suffix"] )
norway_red_pos
```

```python
norway_red_pos["Value_Clean"]=norway_red_pos["Value"].str.replace(r'[^0-9.]', '', regex=True).astype(float)
norway_red_pos["Value_Final"]=norway_red_pos["Value_Clean"]*norway_red_pos["Suffix"]
norway_red_pos=norway_red_pos.reset_index(drop=True)
norway_red_pos=norway_red_pos.iloc[:30]

norway_red_pos["Fund"]=sheet_names[1]
```

```python
base_df=pd.DataFrame()
for i in range(2,len(sheet_names)):
    print(i)
    print(sheet_names[i])
    if sheet_names[i]!="ART":
        df_other=pd.read_excel(excel_file, sheet_name=sheet_names[i])
        df_other
        # Find columns where any cell value contains ")"
        search_string = ")"
        columns_with_char = []

        for col in df_other.columns:
            if df_other[col].astype(str).str.contains(search_string, na=False, regex=False).any():
                columns_with_char.append(col)

        # Get index numbers of those columns
        index_numbers = [df_other.columns.get_loc(col) for col in columns_with_char]
        print(index_numbers)
        other_red_df=df_other[df_other.columns[[index_numbers[0],(index_numbers[0]+2),(index_numbers[0]+5)]]]
        # Find rows where any cell contains the character
        search_string = ")"
        mask = other_red_df.apply(lambda col: col.astype(str).str.contains(search_string, na=False, regex=False), axis=1).any(axis=1)
        other_red_pos = other_red_df[mask]
        other_red_pos.columns=["Rank","Name","Value"]
        other_red_pos["Suffix"]=0
        other_red_pos["Suffix"] = np.where(other_red_pos["Value"].str.contains("BLN"), 10**9,other_red_pos["Suffix"] )
        other_red_pos["Suffix"] = np.where(other_red_pos["Value"].str.contains("MLN"), 10**6,other_red_pos["Suffix"] )
        other_red_pos["Value_Clean"]=other_red_pos["Value"].str.replace(r'[^0-9.]', '', regex=True).astype(float)
        other_red_pos["Value_Final"]=other_red_pos["Value_Clean"]*other_red_pos["Suffix"]
        other_red_pos["Fund"]=sheet_names[i]
        other_red_pos=other_red_pos.reset_index(drop=True)
        other_red_pos=other_red_pos.iloc[:30]
        print(len(other_red_pos))
        base_df=pd.concat([base_df,other_red_pos],ignore_index=True)
    else:
        print("Skipping ART sheet")
    base_df
```

```python
allpos_df=pd.concat([norway_red_pos,base_df],ignore_index=True)
allpos_df
```
