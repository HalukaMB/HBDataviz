```python
data=pd.read_csv("../data/all_pensionfund_red_positions_after.csv",encoding='latin-1', encoding_errors='replace')
data
```

```python
summed_positions=data.groupby("Name").sum().reset_index()

summed_positions_bytotal=summed_positions.sort_values(by='Value_Final', ascending=False)
summed_positions_bytotal=summed_positions_bytotal[["Name","Value_Final"]]
summed_positions_bytotal
```

```python
frequency_positions=data.groupby("Name").count()
frequency_positions=frequency_positions.reset_index()
frequency_positions=frequency_positions[["Name","Fund"]]
frequency_positions.sort_values(by='Fund', ascending=False).head(50)
frequency_positions
```

```python
merged_on_firm_country=pd.merge(data, cleaned_data[["Name","Country"]], on="Name",how="outer",indicator=True)
merged_on_firm_country
```

```python
merged_on_firm_country[merged_on_firm_country['_merge']=="left_only"].to_csv("missing_country_coord.csv", index=False)
```

```python
handfilled=pd.read_csv("missing_country_coord_hand.csv",encoding='latin-1', encoding_errors='replace')
handfilled
```

```python
automaticmatch=merged_on_firm_country[merged_on_firm_country['_merge']=="both"]
automaticmatch
```

```python
automaticmatch["Country"]=automaticmatch["Country"].str.strip()
```

```python
allmatches=pd.concat([automaticmatch, handfilled], ignore_index=True)
allmatches
```

```python
allmatches.groupby("Country").sum().reset_index().sort_values(by='Value_Final', ascending=False)[["Country","Value_Final"]].to_csv("countrywise_total_investments.csv", index=False)
```
