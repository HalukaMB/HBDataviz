```python
import pandas as pd
import requests
import numpy as np
from bs4 import BeautifulSoup
import yfinance
import os
from decimal import Decimal, ROUND_HALF_UP
import locale
from datetime import datetime, timedelta
import json
import time
import io
#https://github.com/patrickloeber/python-github-action-template?tab=readme-ov-file
# Set the locale to English (United States)
locale.setlocale(locale.LC_TIME, 'en_US.UTF-8')
#authorization

```

```python

# Get today's date
end_of_year = "2025-12-31"
end_of_year = datetime.strptime(end_of_year, "%Y-%m-%d")
# Get date 365 days ago
date_365_days_ago = end_of_year - timedelta(days=365)

# Format as string if needed
date_365_days_ago_str = date_365_days_ago.strftime("%Y-%m-%d")

print(date_365_days_ago_str)
```

```python
msci_world = yfinance.download (tickers = "^990100-USD-STRD", start = date_365_days_ago_str, end= end_of_year.strftime("%Y-%m-%d"),
                              interval = "1d")
msci_world
```

```python
emerging_markets = yfinance.download (tickers = "EEM", start = date_365_days_ago_str, end= end_of_year.strftime("%Y-%m-%d"),
                              interval = "1d")
emerging_markets
```

```python
europe_markets = yfinance.download (tickers = "EMUU.L", start = date_365_days_ago_str, end= end_of_year.strftime("%Y-%m-%d"),
                              interval = "1d")
europe_markets
```

```python

eafe_markets = yfinance.download (tickers = "EFA", start = date_365_days_ago_str, end= end_of_year.strftime("%Y-%m-%d"),
                              interval = "1d")
eafe_markets
```

```python
def reduce_index(df,indexname):
    index_red=pd.DataFrame()
    df["Date"]=df.index
    index_red["Date"]=df["Date"]

    index_red[indexname]=df[df.columns[0]]

    index_red=index_red.reset_index(drop=True)
    return index_red
```

```python
europe_markets_red=reduce_index(europe_markets,"EUROPE_MARKETS")
msci_world_red=reduce_index(msci_world,"MSCI_WORLD")
emerging_markets_red=reduce_index(emerging_markets,"EMERGING_MARKETS")
eafe_markets_red=reduce_index(eafe_markets,"EAFE_MARKETS")

europe_markets_red
```

```python
world_emerging=pd.merge(msci_world_red, emerging_markets_red, on="Date", how="outer")
world_emerging=world_emerging.fillna(method="ffill")
world_emerging
```

```python
world_emerging_europe=pd.merge(world_emerging, europe_markets_red, on="Date", how="outer")
world_emerging_europe=world_emerging_europe.fillna(method="ffill")
```

```python
world_emerging_europe_eafe=pd.merge(world_emerging_europe, eafe_markets_red, on="Date", how="outer")
world_emerging_europe_eafe=world_emerging_europe_eafe.fillna(method="ffill")
today_str=end_of_year.strftime("%Y-%m-%d")

world_emerging_europe_eafe_without_today=world_emerging_europe_eafe[world_emerging_europe_eafe["Date"]!=today_str]
world_emerging_europe_eafe_without_today=world_emerging_europe_eafe_without_today.dropna()
world_emerging_europe_eafe_without_today
```

```python
for col in world_emerging_europe_eafe_without_today.columns[1:]:
    world_emerging_europe_eafe_without_today[col]= world_emerging_europe_eafe_without_today[col].apply(lambda x: Decimal(str(x)))

    # Example: Perform precise rounding on the Decimal column
    world_emerging_europe_eafe_without_today[col]= world_emerging_europe_eafe_without_today[col].apply(lambda x: x.quantize(Decimal('1'), rounding=ROUND_HALF_UP))
world_emerging_europe_eafe_without_today
```

```python
def converter_perc(index_name):
    df_perc=pd.DataFrame()
    df_perc["Date"]=world_emerging_europe_eafe_without_today["Date"]
    col_important=world_emerging_europe_eafe_without_today[index_name]
    base_value=col_important.iloc[0]
    df_perc[index_name+"_perc"]=(col_important-base_value)*100/base_value
    df_perc[index_name+"_perc"] = df_perc[index_name+"_perc"].apply(lambda x: float(Decimal(str(x)).quantize(Decimal('0.01'), rounding=ROUND_HALF_UP)))

    return(df_perc)
```

```python
for col in world_emerging_europe_eafe_without_today.columns[1:]:
    print(col)
    df_perc=converter_perc(col)
    print(df_perc)
    world_emerging_europe_eafe_without_today=pd.merge(world_emerging_europe_eafe_without_today,df_perc,on="Date",how="outer")
world_emerging_europe_eafe_without_today
```
