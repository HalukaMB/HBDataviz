```python
import pandas as pd
import glob
import numpy as np
```

```python
vars=[['Country Name','Country Code',"Land",
             'gdp_rel_ger','ppp_rel_ger', 'cost_to_gdp', 'popdensity', 'homicides',
      'FreedomScore', 'PISA Score','timediff_to_germany', 'Cuisine Rank', 
      'SunshineHours_per_year', 'sqm/sqkm', 'max_m', 'min_m', 'English Proficiency', 
       'Anzahl_der_Deutschen_numeric', 'Bureaucracy Score', 'MaxTemperature',
       'MinTemperature', 'LivingConditions', 'Healthcare', 'CostofLiving', 
       'Christian_%', 'Muslim_%', 'Irreligion_%', 'Hindu_%', 'Buddhist_%',
       'Folk religion_%', 'Other religion_%', 'Jewish_%', 'ban',
       'LGBTQ',"Sonnenstunde_pro_Tag"]]
```

```python
data=pd.read_csv("../newdata/all_measures_v4.csv")
data
```

```python
data.drop(columns=['Unnamed: 0',"Code",
      'Country or Territory',], inplace=True)
```

```python
data.columns
```

```python
data_dropped=data.copy()
df_to_check=data_dropped.copy()
```

```python
df_to_check["gdp_rel_ger"]=df_to_check["gdp_rel_ger"]*100
df_to_check["ppp_rel_ger"]=df_to_check["ppp_rel_ger"]*100
df_to_check["Anzahl_der_Deutschen_numeric"]=np.where(df_to_check["Land"]=="Deutschland",np.nan,df_to_check["Anzahl_der_Deutschen_numeric"])
df_to_check
```

```python
df_to_check["FreedomScore"]=pd.to_numeric(df_to_check["FreedomScore"], errors='coerce')
df_to_check["LGBTQ"]=pd.to_numeric(df_to_check["LGBTQ"], errors='coerce')
df_to_check
```

```python
df_to_check["SunshineHours_per_year"]=pd.to_numeric(df_to_check["SunshineHours_per_year"], errors='coerce')
df_to_check["Sonnenstunde_pro_Tag"]=df_to_check["SunshineHours_per_year"]/365
df_to_check
```

```python
df_to_check["Mehrals5Morde_pro_100k_boolean"]=np.where(df_to_check["homicides"]>=5,True,False)
df_to_check
```

```python
df_to_check["Mehrals1000Deutsche_boolean"]=np.where(df_to_check["Anzahl_der_Deutschen_numeric"]>=1000,True,False)
df_to_check
```

```python
df_to_check["Alkoholverbot_boolean"]=np.where(df_to_check["ban"]>=1,True,False)
df_to_check
```

```python
df_to_check["Kueste_boolean"]=np.where(df_to_check['sqm/sqkm']>0,True,False)
df_to_check
```

```python
df_to_check["Christian_boolean"]=np.where(df_to_check["Christian_%"]>1,True,False)
df_to_check["Muslim_boolean"]=np.where(df_to_check["Muslim_%"]>1,True,False)
df_to_check["Jewish_boolean"]=np.where(df_to_check["Jewish_%"]>1,True,False)
df_to_check
```

```python
df_to_check["English_boolean"]=np.where(df_to_check["English Proficiency"]>500,True,False)
df_to_check["Cuisine_boolean"]=np.where(df_to_check["Cuisine Rank"]>0,True,False)
df_to_check.sort_values("English Proficiency",ascending=False)
```
